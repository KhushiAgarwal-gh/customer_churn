```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('../data/churn.csv')
df.head()
```

```python
df.info(); df.isna().sum()
```

```python
df['Churn'].value_counts(normalize=True).plot.bar(title='Churn rate'); plt.show()
```

```python
pd.crosstab(df['Contract'], df['Churn'], normalize='index').plot.bar(stacked=True, title='Churn by Contract'); plt.show()
```

```python
df.boxplot(column='tenure', by='Churn'); plt.show()
```

```python
df.boxplot(column='MonthlyCharges', by='Churn'); plt.show()
```

```python
pd.crosstab(df['SupportCalls'], df['Churn'], normalize='index')['Yes'].plot(title='Churn rate vs Support calls'); plt.show()
```
